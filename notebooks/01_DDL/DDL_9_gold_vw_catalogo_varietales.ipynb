```python
%sql
CREATE OR REPLACE VIEW vitivinicola_catalog.gold.vw_catalogo_varietales AS
SELECT
    t.tienda,
    t.pais,
    t.moneda,
    p.titulo,
    p.bodega,
    p.categoria,
    v.varietal,
    f.precio,
    f.precio_tachado,
    f.stock
FROM vitivinicola_catalog.gold.fact_catalogo f
INNER JOIN vitivinicola_catalog.gold.dim_productos p
    ON f.sk_producto = p.sk_producto
INNER JOIN vitivinicola_catalog.gold.dim_tiendas t
    ON f.sk_tienda = t.sk_tienda
LEFT JOIN vitivinicola_catalog.gold.dim_producto_varietal pv
    ON p.sk_producto = pv.sk_producto
LEFT JOIN vitivinicola_catalog.gold.dim_varietales v
    ON pv.sk_varietal = v.sk_varietal;
```
