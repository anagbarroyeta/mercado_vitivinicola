```python
%sql
-- CREACIÓN CATALOGO
CREATE CATALOG IF NOT EXISTS vitivinicola_catalog;

SHOW CATALOGS;
```

```python

%sql
-- CREACIÓN SCHEMAS 
USE CATALOG vitivinicola_catalog;

-- LANDING 
CREATE SCHEMA IF NOT EXISTS vitivinicola_catalog.landing
COMMENT 'Esquema para archivos crudos en volumenes';

-- BRONZE 
CREATE SCHEMA IF NOT EXISTS vitivinicola_catalog.bronze
COMMENT 'Capa Bronze: Datos crudos sin procesar directos del origen';

-- SILVER
CREATE SCHEMA IF NOT EXISTS vitivinicola_catalog.silver
COMMENT 'Capa Silver: Datos limpios, validados y con tipos correctos';

-- GOLD
CREATE SCHEMA IF NOT EXISTS vitivinicola_catalog.gold  
COMMENT 'Capa Gold: Lista para consumo';

-- VERIFICAR
SHOW SCHEMAS;
```

```python
%sql
-- CREACIÓN VOLUME MANAGED 
CREATE VOLUME IF NOT EXISTS vitivinicola_catalog.landing.archivos
COMMENT 'Volume para almacenar archivos CSV crudos';
 
SHOW VOLUMES IN vitivinicola_catalog.landing;
```
