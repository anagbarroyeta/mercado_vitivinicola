```python
%sql

CREATE TABLE IF NOT EXISTS vitivinicola_catalog.gold.dim_productos (
    sk_producto BIGINT GENERATED ALWAYS AS IDENTITY,
    bk_producto BIGINT COMMENT 'Business Key original (ID)',
    titulo STRING,
    bodega STRING,
    categoria STRING
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Dimensión de productos vitivinícolas';
```
