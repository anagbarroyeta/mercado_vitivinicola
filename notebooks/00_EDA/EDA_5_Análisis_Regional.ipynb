```python
%sql

-- TOP 3 BODEGAS POR SEGMENTO DE PRECIO
 
WITH ranking_bodegas AS (

    SELECT
        segmento_precio,
        bodega,
        COUNT(*) AS cantidad_productos,
        ROW_NUMBER() OVER (PARTITION BY segmento_precio ORDER BY COUNT(*) DESC) AS ranking

    FROM wine_catalog.bronze.precios_regional 

    GROUP BY
        segmento_precio,
        bodega
)

SELECT
    segmento_precio,
    bodega,
    cantidad_productos

FROM ranking_bodegas

WHERE ranking <= 3

ORDER BY
    segmento_precio, cantidad_productos DESC;

-- Norton presenta mayor presencia en el segmento económico. Susana Balbo concentra productos en segmentos medio y
-- premium.
```

```python
%sql

-- STOCK POR SEGMENTO

SELECT
    segmento_precio,
    COUNT(*) AS productos_agotados
FROM wine_catalog.bronze.precios_regional 
WHERE stock = false
GROUP BY segmento_precio
ORDER BY productos_agotados DESC;

-- La ausencia de stock puede sugerir alta rotación o reposición insuficiente, pero no puede interpretarse 
-- directamente como volumen de ventas.
```
