```python
%sql

-- TAMAÑO DEL DATASET 
                                                
SELECT COUNT(*) as total_registros
FROM vitivinicola_catalog.bronze.products;
```

```python
%sql
    
-- ESTRUCTURA DEL DATASET

DESCRIBE TABLE vitivinicola_catalog.bronze.products;
```

```python
%sql

-- MUESTRA DE DATOS

SELECT 
    pais,
    moneda,
    id,
    titulo,
    bodega,
    tipo,
    tags,
    precio,
    precio_tachado,
    stock,
    publicado,
    descripcion,
    _rescued_data
FROM vitivinicola_catalog.bronze.products
LIMIT 10;
```

```python
%sql
-- DISTRIBUCIÓN DE VALORES NULOS EN ATRIBUTOS
-- Se utiliza la función RLIKE para filtrar los ID corruptos

WITH total AS (
    SELECT COUNT(*) as total 
    FROM vitivinicola_catalog.bronze.products
    where id RLIKE '^[0-9]+$' 
),
nulos AS (
    SELECT
        COUNT(*) - COUNT(id) as nulos_id,
        COUNT(*) - COUNT(titulo) as nulos_titulo,
        COUNT(*) - COUNT(bodega) as nulos_bodega,
        COUNT(*) - COUNT(tipo) as nulos_tipo,
        COUNT(*) - COUNT(tags) as nulos_tags,
        COUNT(*) - COUNT(precio) as nulos_precio,
        COUNT(*) - COUNT(precio_tachado) as nulos_precio_tachado,
        COUNT(*) - COUNT(stock) as nulos_stock,
        COUNT(*) - COUNT(publicado) as nulos_publicado,
        COUNT(*) - COUNT(descripcion) as nulos_descripcion,
        COUNT(*) - COUNT(pais) as nulos_pais,
        COUNT(*) - COUNT(moneda) as nulos_moneda
    FROM vitivinicola_catalog.bronze.products
    where id RLIKE '^[0-9]+$'
)
SELECT 
    t.total as total_registros,
    n.*
FROM total t, nulos n;

-- La columna tipo concentra la mayor cantidad de valores nulos con más de mil registros sin clasificar.

```

```python
%sql

-- CANTIDAD DE REGISTROS DUPLICADOS

SELECT
    id,
    COUNT(*) AS repeticiones

FROM vitivinicola_catalog.bronze.products
where id RLIKE '^[0-9]+$'
GROUP BY id
HAVING COUNT(*) > 1;
```
