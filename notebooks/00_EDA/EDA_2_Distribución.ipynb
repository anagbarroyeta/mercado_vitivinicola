```python
%sql
-- RECUENTO VALORES ÚNICOS (CARDINALIDAD)
-- Se utiliza la función RLIKE para filtrar los ID corruptos

SELECT 
    COUNT(DISTINCT id)      as productos_unicos,
    COUNT(DISTINCT bodega)  as bodegas_unicas,
    COUNT(DISTINCT tipo)    as tipos_unicos,
    COUNT(DISTINCT moneda)  as monedas_unicas,
    COUNT(DISTINCT pais)    as paises_unicos
FROM wine_catalog.bronze.products
WHERE id RLIKE '^[0-9]+$';
```

```python
%sql
-- DISTRIBUCIÓN POR MONEDA

SELECT moneda, COUNT(*) as cantidad,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER(), 2) as pct
FROM wine_catalog.bronze.products
WHERE id RLIKE '^[0-9]+$'
GROUP BY moneda 
ORDER BY cantidad DESC;
```

```python
%sql
-- DISTRIBUCIÓN POR BODEGA

SELECT bodega, COUNT(*) as cantidad,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER(), 2) as pct
FROM wine_catalog.bronze.products
WHERE id RLIKE '^[0-9]+$'
GROUP BY bodega 
ORDER BY cantidad DESC;
```

```python
%sql
-- DISTRIBUCIÓN POR TIPO

SELECT tipo, COUNT(*) as cantidad,
    ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER(), 2) as pct
FROM wine_catalog.bronze.products
WHERE id RLIKE '^[0-9]+$'
GROUP BY tipo 
ORDER BY cantidad DESC;
```
