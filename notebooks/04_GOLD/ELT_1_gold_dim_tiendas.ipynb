```python
%sql
-- DIMENSION TIENDAS
MERGE INTO vitivinicola_catalog.gold.dim_tiendas AS target
USING (SELECT DISTINCT tienda, pais, moneda FROM vitivinicola_catalog.silver.productos) AS source
ON target.tienda = source.tienda
WHEN MATCHED THEN
  UPDATE SET target.pais = source.pais, target.moneda = source.moneda
WHEN NOT MATCHED THEN
  INSERT (tienda, pais, moneda) VALUES (source.tienda, source.pais, source.moneda);
```
