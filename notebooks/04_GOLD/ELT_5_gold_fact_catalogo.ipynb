```python
%sql
-- FACT PRODUCTOS Y PRECIOS
INSERT OVERWRITE vitivinicola_catalog.gold.fact_catalogo
SELECT 
    dp.sk_producto,
    dt.sk_tienda,
    s.precio,
    s.precio_tachado,
    s.stock
FROM vitivinicola_catalog.silver.productos s
INNER JOIN vitivinicola_catalog.gold.dim_productos dp 
    ON s.id = dp.bk_producto
INNER JOIN vitivinicola_catalog.gold.dim_tiendas dt 
    ON s.tienda = dt.tienda;
```
