```python
%sql
-- DIMENSIÓN PRODUCTOS
MERGE INTO vitivinicola_catalog.gold.dim_productos AS target
USING vitivinicola_catalog.silver.productos AS source
ON target.bk_producto = source.id
WHEN MATCHED THEN
  UPDATE SET target.titulo = source.titulo, target.bodega = source.bodega, target.categoria = source.categoria
WHEN NOT MATCHED THEN
  INSERT (bk_producto, titulo, bodega, categoria) VALUES (source.id, source.titulo, source.bodega, source.categoria);

```
