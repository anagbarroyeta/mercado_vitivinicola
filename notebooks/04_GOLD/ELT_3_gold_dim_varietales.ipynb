```python
%sql
-- DIMENSIÓN DE VARIETALES
MERGE INTO vitivinicola_catalog.gold.dim_varietales AS target
USING (
    SELECT DISTINCT 
        CASE 
            WHEN LOWER(TRIM(tag_crudo)) = 'malbec' THEN 'Malbec'
            WHEN LOWER(TRIM(tag_crudo)) IN ('cabernetsauvignon', 'cabernet sauvignon') THEN 'Cabernet Sauvignon'
            WHEN LOWER(TRIM(tag_crudo)) = 'chardonnay' THEN 'Chardonnay'
            WHEN LOWER(TRIM(tag_crudo)) IN ('pinot noir', 'pinotnoir') THEN 'Pinot Noir'
            WHEN LOWER(TRIM(tag_crudo)) IN ('cabernet franc', 'cabernetfranc') THEN 'Cabernet Franc'
            WHEN LOWER(TRIM(tag_crudo)) IN ('torrontã©s', 'torrontes') THEN 'Torrontés'
            WHEN LOWER(TRIM(tag_crudo)) = 'tempranillo' THEN 'Tempranillo'
            WHEN LOWER(TRIM(tag_crudo)) = 'verdejo' THEN 'Verdejo'
            WHEN LOWER(TRIM(tag_crudo)) = 'merlot' THEN 'Merlot'
            WHEN LOWER(TRIM(tag_crudo)) = 'syrah' THEN 'Syrah'
            WHEN LOWER(TRIM(tag_crudo)) = 'cinsault' THEN 'Cinsault'
            WHEN LOWER(TRIM(tag_crudo)) = 'tannat' THEN 'Tannat'
            WHEN LOWER(TRIM(tag_crudo)) = 'macabeo' THEN 'Macabeo'
            WHEN LOWER(TRIM(tag_crudo)) = 'sangiovese' THEN 'Sangiovese'
            WHEN LOWER(TRIM(tag_crudo)) IN ('albariã±o', 'albarino') THEN 'Albariño'
            WHEN LOWER(TRIM(tag_crudo)) = 'blend' THEN 'Blend'
            WHEN LOWER(TRIM(tag_crudo)) IN ('champagne', 'espumantes', 'espumante', 'espumosos') THEN 'Champagne'
            WHEN LOWER(TRIM(tag_crudo)) = 'pinot gris' THEN 'Pinot Gris'
            WHEN LOWER(TRIM(tag_crudo)) = 'pinot grigio' THEN 'Pinot Grigio'
            WHEN LOWER(TRIM(tag_crudo)) = 'chenin' THEN 'Chenin Blanc'
            WHEN LOWER(TRIM(tag_crudo)) = 'semillon' THEN 'Semillon'
            WHEN LOWER(TRIM(tag_crudo)) = 'barbera' THEN 'Barbera'
            WHEN LOWER(TRIM(tag_crudo)) = 'nebbiolo' THEN 'Nebbiolo'
            WHEN LOWER(TRIM(tag_crudo)) = 'carignan' THEN 'Carignan'
            WHEN LOWER(TRIM(tag_crudo)) = 'garnacha' THEN 'Garnacha'
            WHEN LOWER(TRIM(tag_crudo)) = 'bonarda' THEN 'Bonarda'
            WHEN LOWER(TRIM(tag_crudo)) = 'viognier' THEN 'Viognier'
            WHEN LOWER(TRIM(tag_crudo)) IN ('rosados', 'rosado', 'rosã©') THEN 'Rosado'
            WHEN LOWER(TRIM(tag_crudo)) = 'riesling' THEN 'Riesling'
            WHEN LOWER(TRIM(tag_crudo)) IN ('tintos', 'tinto') THEN 'Otros Tintos'
            WHEN LOWER(TRIM(tag_crudo)) IN ('vermut', 'vermouth', 'vermú') THEN 'Vermut'
            ELSE 'Otros'
        END AS varietal_limpio
    FROM (
        SELECT tag_crudo
        FROM vitivinicola_catalog.silver.productos
        LATERAL VIEW EXPLODE(SPLIT(tags, ',')) AS tag_crudo
        WHERE tags IS NOT NULL
    )
) AS source
ON target.varietal = source.varietal_limpio

WHEN NOT MATCHED THEN
  INSERT (varietal) VALUES (source.varietal_limpio);
```
