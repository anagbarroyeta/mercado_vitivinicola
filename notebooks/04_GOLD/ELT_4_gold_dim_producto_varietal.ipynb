```python
%sql
-- TABLA PUENTE 
INSERT OVERWRITE vitivinicola_catalog.gold.dim_producto_varietal
SELECT DISTINCT
    dp.sk_producto,
    dv.sk_varietal
FROM (
    -- Explotar tags
    SELECT id, TRIM(tag_crudo) AS tag_limpio
    FROM vitivinicola_catalog.silver.productos
    LATERAL VIEW EXPLODE(SPLIT(tags, ',')) AS tag_crudo
    WHERE tags IS NOT NULL
) s
-- Cruzar con dim_productos
INNER JOIN vitivinicola_catalog.gold.dim_productos dp 
    ON s.id = dp.bk_producto
-- Cruzar con dim_varietales 
INNER JOIN vitivinicola_catalog.gold.dim_varietales dv 
    ON dv.varietal = CASE 
        WHEN LOWER(s.tag_limpio) = 'malbec' THEN 'Malbec'
        WHEN LOWER(s.tag_limpio) IN ('cabernetsauvignon', 'cabernet sauvignon') THEN 'Cabernet Sauvignon'
        WHEN LOWER(s.tag_limpio) = 'chardonnay' THEN 'Chardonnay'
        WHEN LOWER(s.tag_limpio) IN ('pinot noir', 'pinotnoir') THEN 'Pinot Noir'
        WHEN LOWER(s.tag_limpio) IN ('cabernet franc', 'cabernetfranc') THEN 'Cabernet Franc'
        WHEN LOWER(s.tag_limpio) IN ('torrontã©s', 'torrontes') THEN 'Torrontés'
        WHEN LOWER(s.tag_limpio) = 'tempranillo' THEN 'Tempranillo'
        WHEN LOWER(s.tag_limpio) = 'verdejo' THEN 'Verdejo'
        WHEN LOWER(s.tag_limpio) = 'merlot' THEN 'Merlot'
        WHEN LOWER(s.tag_limpio) = 'syrah' THEN 'Syrah'
        WHEN LOWER(s.tag_limpio) = 'cinsault' THEN 'Cinsault'
        WHEN LOWER(s.tag_limpio) = 'tannat' THEN 'Tannat'
        WHEN LOWER(s.tag_limpio) = 'macabeo' THEN 'Macabeo'
        WHEN LOWER(s.tag_limpio) = 'sangiovese' THEN 'Sangiovese'
        WHEN LOWER(s.tag_limpio) IN ('albariã±o', 'albarino') THEN 'Albariño'
        WHEN LOWER(s.tag_limpio) = 'blend' THEN 'Blend'
        WHEN LOWER(s.tag_limpio) IN ('champagne', 'espumantes', 'espumante', 'espumosos') THEN 'Champagne'
        WHEN LOWER(s.tag_limpio) = 'pinot gris' THEN 'Pinot Gris'
        WHEN LOWER(s.tag_limpio) = 'pinot grigio' THEN 'Pinot Grigio'
        WHEN LOWER(s.tag_limpio) = 'chenin' THEN 'Chenin Blanc'
        WHEN LOWER(s.tag_limpio) = 'semillon' THEN 'Semillon'
        WHEN LOWER(s.tag_limpio) = 'barbera' THEN 'Barbera'
        WHEN LOWER(s.tag_limpio) = 'nebbiolo' THEN 'Nebbiolo'
        WHEN LOWER(s.tag_limpio) = 'carignan' THEN 'Carignan'
        WHEN LOWER(s.tag_limpio) = 'garnacha' THEN 'Garnacha'
        WHEN LOWER(s.tag_limpio) = 'bonarda' THEN 'Bonarda'
        WHEN LOWER(s.tag_limpio) = 'viognier' THEN 'Viognier'
        WHEN LOWER(s.tag_limpio) IN ('rosados', 'rosado', 'rosã©') THEN 'Rosado'
        WHEN LOWER(s.tag_limpio) = 'riesling' THEN 'Riesling'
        WHEN LOWER(s.tag_limpio) IN ('tintos', 'tinto') THEN 'Otros Tintos'
        WHEN LOWER(s.tag_limpio) IN ('vermut', 'vermouth', 'vermú') THEN 'Vermut'
        ELSE 'Otros'
    END;
```
